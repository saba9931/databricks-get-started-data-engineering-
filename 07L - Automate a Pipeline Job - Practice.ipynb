### Task 1: Compare manual vs. automated Bronze tables

```python
%sql
USE CATALOG dbacademy;
USE SCHEMA get_started_de;

SELECT 'Manual (Lesson 6)' AS source, COUNT(*) AS rows FROM current_employees_bronze
UNION ALL
SELECT 'Job (Lesson 7)', COUNT(*) FROM current_employees_bronze_job;
```
